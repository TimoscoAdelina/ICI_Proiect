# Analiza comparativă România vs UE privind indicatorii climatici

## Obiectiv

Scopul acestei analize este de a compara poziția României față de media Uniunii Europene 
din perspectiva unor indicatori relevanți pentru tranziția climatică.

Graficul ilustrează traiectoria planificată de reducere a emisiilor de gaze cu efect de seră la nivelul Uniunii Europene. Se observă un obiectiv ambițios de reducere semnificativă până în anul 2030, urmat de atingerea neutralității climatice până în 2050. Această evoluție reflectă angajamentul UE de a deveni lider global în combaterea schimbărilor climatice.

Germania este una dintre țările europene care au un rol central în promovarea și finanțarea inițiativelor climatice. ### Suedia — lider legislativ în politici climatice

Suedia este una dintre primele țări care au implementat politici climatice bazate pe instrumente economice și legislative coerente.

Un element central îl reprezintă introducerea taxei pe carbon în 1991, una dintre primele de acest tip la nivel global. În prezent, Suedia aplică una dintre cele mai ridicate taxe pe carbon, contribuind semnificativ la reducerea emisiilor de CO₂.

În 2017, Suedia a adoptat un cadru legislativ climatic care stabilește obiective pe termen lung, inclusiv atingerea neutralității climatice până în 2045. Acest cadru este susținut de instituții independente care monitorizează implementarea politicilor.

Eficiența acestor măsuri este reflectată în reducerea emisiilor per capita și a intensității carbonice a economiei, ambele situându-se sub media Uniunii Europene. De asemenea, sistemele de taxare a carbonului și EU ETS acoperă majoritatea emisiilor provenite din combustibili fosili.

Aceste elemente poziționează Suedia ca un exemplu de bună practică în domeniul politicilor climatice la nivel european.

## Bune practici identificate în orașele europene

O primă bună practică este abordarea integrată. Documentele Comisiei Europene arată că neutralitatea climatică nu poate fi atinsă prin măsuri sectoriale separate, ci printr-o strategie holistică care combină mobilitatea, energia, clădirile și implicarea cetățenilor. Climate City Contracts sunt relevante deoarece reunesc într-un singur cadru angajamentele politice, planul de acțiune și planul de investiții.

A doua bună practică este guvernanța colaborativă. Exemplele promovate de NetZeroCities includ orașe precum Oslo și Leuven, unde administrația colaborează cu sectorul privat, universități și comunități locale. În cazul Leuven, contractul climatic implică peste 25 de actori locali, ceea ce facilitează implementarea.

A treia bună practică este pregătirea financiară. Orașele performante sunt cele care reușesc să transforme proiectele în investiții „bankable”, atrăgând atât fonduri publice, cât și private. Aceasta presupune expertiză financiară și o planificare riguroasă.

A patra bună practică este implicarea cetățenilor. Curtea de Conturi Europeană arată că aproximativ 75% dintre proiecte au întâmpinat dificultăți din cauza participării insuficiente a populației.

Aceste bune practici evidențiază faptul că succesul nu este determinat de tehnologie, ci de modul în care aceasta este integrată în sistemul urban și acceptată social.

Se poate afirma că orașele care tratează tranziția climatică ca un proces colectiv și interdisciplinar au șanse semnificativ mai mari de reușită.


## Lecții învățate și greșeli care trebuie evitate

Una dintre cele mai importante greșeli este tehnologizarea fără guvernanță. Conform OECD (2023), multe proiecte smart city au probleme de scalabilitate și sustenabilitate deoarece nu sunt clar definite responsabilitățile instituționale și regulile privind utilizarea datelor.

Un exemplu relevant este proiectul Sidewalk Toronto, abandonat în 2020. Inițiativa, dezvoltată de Sidewalk Labs (parte a Alphabet), urmărea transformarea unei zone urbane din Toronto într-un model global de smart city, bazat pe utilizarea extensivă a datelor, senzorilor și inteligenței artificiale. Cu toate acestea, proiectul a generat controverse majore legate de guvernanța datelor, lipsa clarității privind cine controlează informațiile colectate și modul în care acestea sunt utilizate. De asemenea, au existat preocupări privind supravegherea excesivă și lipsa unui control democratic clar. În final, opoziția publică, combinată cu incertitudinile economice și instituționale, a condus la abandonarea proiectului. Acest caz demonstrează că, în absența legitimității sociale și a unui cadru de guvernanță transparent, chiar și cele mai avansate inițiative tehnologice pot eșua.

O altă greșeală este pilotarea fără replicare. Auditul Curții de Conturi Europene arată că multe proiecte din programul Lighthouse au produs rezultate pozitive la scară locală, însă nu au fost scalate eficient sau replicate în alte orașe. Lipsa standardizării, a unor modele financiare sustenabile și a coordonării între niveluri de guvernare limitează impactul acestor inițiative.

În plus, subestimarea birocrației și a capacității administrative reprezintă un obstacol major. Majoritatea barierelor identificate sunt instituționale, nu tehnice.În multe cazuri, problema nu este lipsa finanțării, ci incapacitatea de a gestiona și implementa eficient proiectele existente.

Aceste exemple demonstrează că eșecul nu apare din lipsa tehnologiei, ci din lipsa unei structuri clare de implementare și acceptare socială.

Rezultă că orice strategie de smart city trebuie să prioritizeze guvernanța, transparența și implicarea cetățenilor înaintea implementării tehnologice.

### Poziția României în context european

România se situează într-o poziție intermediară în cadrul tranziției verzi europene, caracterizată prin performanțe relativ bune la nivelul unor indicatori cantitativi, dar cu decalaje semnificative în ceea ce privește eficiența sistemică și implementarea politicilor de mediu. 

La nivel european, inițiativele de tip „climate-neutral cities” și strategiile de economie circulară sunt susținute de state lider precum Germania, Olanda sau Suedia, care combină investițiile în infrastructură verde cu sisteme eficiente de guvernanță și monitorizare a datelor. În acest context, România participă limitat la inițiativele europene de vârf, având doar trei entități selectate în cadrul EU Mission Cities, ceea ce indică un nivel redus de integrare în nucleul inovator al politicilor urbane climatice.

Din perspectiva indicatorilor de mediu, România prezintă un profil contrastant: pe de o parte, nivelul emisiilor de gaze cu efect de seră per capita și cantitatea de deșeuri generate sunt sub media Uniunii Europene; pe de altă parte, eficiența utilizării resurselor, ratele de reciclare și intensitatea emisiilor raportată la PIB rămân semnificativ sub standardele europene. 

Această combinație sugerează că performanța aparent favorabilă este determinată mai degrabă de un nivel mai scăzut al consumului și dezvoltării economice, decât de eficiența politicilor de mediu, ceea ce evidențiază necesitatea unor investiții structurale și a unei guvernanțe mai coerente în domeniul sustenabilității.

```python
import pandas as pd
from pathlib import Path

# setare paths
BASE_DIR = Path.cwd().resolve()
PROJECT_ROOT = BASE_DIR.parent

romania_file = PROJECT_ROOT / "data/raw/romania_emissions.xlsx"
eu_file = PROJECT_ROOT / "data/raw/eu_emissions.xlsx"

# citire
df_ro = pd.read_excel(romania_file, skiprows=8)
df_eu = pd.read_excel(eu_file, skiprows=8)

df_ro.head()
```

```python
df_ro.columns = df_ro.columns.str.strip()
df_eu.columns = df_eu.columns.str.strip()
```

```python
df_ro = df_ro.loc[:, ~df_ro.columns.str.contains("Unnamed")]
df_eu = df_eu.loc[:, ~df_eu.columns.str.contains("Unnamed")]
```

```python
df_ro = df_ro.rename(columns={"TIME": "gas"})
df_eu = df_eu.rename(columns={"TIME": "gas"})
```

```python
df_ro_long = df_ro.melt(id_vars="gas", var_name="year", value_name="value")
df_ro_long["region"] = "Romania"

df_eu_long = df_eu.melt(id_vars="gas", var_name="year", value_name="value")
df_eu_long["region"] = "EU27"
```

```python
df = pd.concat([df_ro_long, df_eu_long])
df["year"] = df["year"].astype(int)

```

```python
output_file = PROJECT_ROOT / "data" / "processed" / "emissions_clean.csv"
output_file.parent.mkdir(parents=True, exist_ok=True)

df.to_csv(output_file, index=False)
print(f"Saved to: {output_file}")
```

```python
df = df[df["gas"] != "GEO (Labels)"]
df = df.dropna(subset=["value"])
```

```python
df = df[df["gas"].isin(["CO2", "CH4", "N2O"])]
```

```python
df_co2 = df[df["gas"] == "CO2"].copy()
df_co2 = df_co2.sort_values(["region", "year"])

df_co2["index_2015"] = (
    df_co2.groupby("region")["value"]
          .transform(lambda x: x / x.iloc[0] * 100)
)

df_co2
```

```python
from pathlib import Path

figures_dir = Path("../outputs/figures")
figures_dir.mkdir(parents=True, exist_ok=True)
plt.figure(figsize=(10, 6))
sns.lineplot(data=df_co2, x="year", y="index_2015", hue="region", marker="o")

plt.title("CO2 emissions index (2015 = 100): Romania vs EU")
plt.xlabel("Year")
plt.ylabel("Index (2015 = 100)")
plt.legend(title="Region")
plt.tight_layout()
plt.savefig(figures_dir / "co2_index_romania_vs_eu.png", dpi=300, bbox_inches="tight")
plt.show()
```

Pentru a compara corect România cu UE, emisiile au fost transformate într-un index cu baza 2015 = 100. Aceasta înseamnă că toate valorile sunt raportate la nivelul din 2015: o valoare de 90 indică o reducere de 10%, iar o valoare de 110 indică o creștere de 10% față de anul de referință.

## Interpretare: Emisii CO₂ — România vs UE27 (2015–2023)

> **Notă metodologică:** Graficul prezintă un **indice relativ (2015 = 100)**, 
> nu valori absolute. Aceasta înseamnă că ambele serii pornesc din același punct 
> și reflectă **ritmul de schimbare** față de anul de referință, nu scala emisiilor 
> în tone. UE27 emite în total de ~40x mai mult decât România — graficul nu arată acest lucru.

### Ce observăm:

**1. UE27 — reducere mai rapidă și mai constantă**
Media europeană a scăzut consistent de la 100 (2015) la ~80 (2023),
o reducere de aproximativ **20 puncte** în 8 ani, fără oscilații majore.
Aceasta reflectă politici climatice coordonate și investiții susținute în energie verde.

**2. România — traiectorie mai volatilă**
România oscilează mai mult între ani, sugerând că reducerile de emisii 
sunt mai degrabă **conjuncturale** (criză energetică, scădere industrială) 
decât rezultatul unor politici climatice structurale.

**3. 2020 — efectul COVID vizibil la ambele**
Scăderea bruscă din 2020 este comună ambelor serii și reflectă 
oprirea activității industriale și reducerea transportului în perioada pandemiei.

**4. 2021 — revenire post-pandemică**
Ambele înregistrează o ușoară creștere în 2021, odată cu reluarea activității economice.

**5. 2022–2023 — România scade mai rapid decât UE**
România ajunge sub media UE ca ritm de scădere — parțial explicat prin 
reducerea producției industriale și închiderea unor capacități pe cărbune.

### Concluzie
Deși România pare să reducă emisiile în ritm similar cu UE, **natura acestor reduceri 
diferă fundamental**: UE investește activ în tranziție verde, în timp ce România 
beneficiază mai mult de factori externi (criză energetică, dezindustrializare).
Fără politici climatice ambițioase, România riscă să rămână în urmă față de 
țintele europene pentru **2030**.

Domeniul 1: Mediul Construit
### Principiu-cheie: Renovare Profundă → Standard nZEB → Standard ZEB

**Ce vizează misiunea:**
- 75% din clădirile UE sunt energetic ineficiente; sectorul contribuie cu >⅓ din emisiile GES legate de energie
- Rata actuală de renovare: **<1%/an** → la această rată, neutralitatea climatică s-ar atinge în ~100 de ani
- **Obiectiv M100:** **3–4% renovări aprofundate/an**, astfel încât **90% din fondul construit** să fie renovat sau demolat până în **2030**
- **Standard de referință:**
  - Clădiri noi → **nZEB** (consum aproape zero, min. 30% din surse regenerabile) → **ZEB** din 2030
  - Clădiri existente publice → nZEB obligatoriu din 2028, privat din 2030

**Decalajul România vs. UE:** Clădirile rezidențiale din România au un consum specific mediu cu ~40% mai mare față de media UE și o rată de renovare aprofundată sub 0,5%/an.

```python
# ─── Traictoria de renovare: scenariul actual vs. scenariul M100 ───
ani = list(range(2024, 2051))

# Scenariul actual: rata de renovare ~0.8%/an
fond_nereabilitat_actual = [100 * (1 - 0.008) ** (a - 2024) for a in ani]

# Scenariul M100-UE: 3.5%/an, neutralitate ~2035-2038
fond_nereabilitat_m100 = [max(0, 100 * (1 - 0.035) ** (a - 2024)) for a in ani]

# Scenariul M100-Romania: 2.5%/an, neutralitate ~2045
fond_nereabilitat_ro = [max(0, 100 * (1 - 0.025) ** (a - 2024)) for a in ani]

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=ani, y=fond_nereabilitat_actual,
    name='Trend actual (0.8%/an)',
    line=dict(color='#B71C1C', width=2, dash='dot'),
    fill='tozeroy', fillcolor='rgba(183,28,28,0.05)'
))
fig.add_trace(go.Scatter(
    x=ani, y=fond_nereabilitat_ro,
    name='Scenariul M100-România (2.5%/an)',
    line=dict(color='#CC0000', width=2.5),
    fill='tozeroy', fillcolor='rgba(204,0,0,0.07)'
))
fig.add_trace(go.Scatter(
    x=ani, y=fond_nereabilitat_m100,
    name='Obiectiv M100-UE (3.5%/an)',
    line=dict(color='#003399', width=2.5),
    fill='tozeroy', fillcolor='rgba(0,51,153,0.07)'
))

fig.add_vline(x=2030, line_dash='dash', line_color='#003399',
              annotation_text='Target UE 2030', annotation_position='top left', annotation_yshift=15, annotation_xshift=5)
fig.add_vline(x=2035, line_dash='dash', line_color='#CC0000',
              annotation_text='Target RO 2035', annotation_position='top right', annotation_yshift =15, annotation_xshift=10)
fig.add_hline(y=10, line_dash='dot', line_color='#2E7D32',
              annotation_text='Prag neutralitate (10% fond nereabilitat)')

fig.update_layout(
    title='Fondul de Clădiri Nereabilitate<br>Scenarii de Renovare 2024–2050',
    title_x=0.5,
    xaxis_title='An',
    yaxis_title='% Fond Construit Nereabilitat',
    height=380,
    width=850,
    legend=dict(x=0.65, y=0.95)
    # margin =dict(t=100)
)
fig.show()
```

**Planificare Urbană — Măsuri M100:**
- **Transit-Oriented Development (TOD)** — densificare în jurul nodurilor de transport public
- **Principiul proximității** — servicii esențiale la max. 15 minute (pe jos/bicicletă)
- **Coordonare metropolitană** — cooperare funcțională între municipiu și zona periurbană
- **SECAP** (Sustainable Energy and Climate Action Plans) — obligatorii pentru semnatarii Covenant of Mayors
- **Masterplan de regenerare urbană** — reconversia terenurilor industriale abandonate (brownfields)

**Digitalizare & Date Urbane — Măsuri M100:**
1. **Platformă urbană de date** — agregare date din surse multiple (IoT, cadastru, transport, energie)
2. **Digital Twin** al orașului — model digital 3D actualizat în timp real
3. **Obligații de raportare** emisii net-zero (Scop 1, 2, 3) conform standardelor europene
