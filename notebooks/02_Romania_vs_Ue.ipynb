# Analiza comparativă România vs UE privind indicatorii climatici

## Obiectiv
Scopul acestei analize este de a compara poziția României față de media Uniunii Europene din perspectiva unor indicatori relevanți pentru tranziția climatică.

## Context general: Inițiative climatice în Uniunea Europeană

Uniunea Europeană se poziționează ca unul dintre liderii globali în domeniul politicilor climatice, prin implementarea unor strategii integrate menite să reducă impactul asupra mediului și să accelereze tranziția către o economie sustenabilă.

Aceste inițiative urmăresc realizarea unei tranziții ecologice care să fie nu doar eficientă din punct de vedere economic, ci și echitabilă din punct de vedere social, protejând atât mediul, cât și cetățenii.

Sursa: climate.ec.europa.eu

```python

years = ["1990", "2030", "2050"]
emissions = [100, 45, 0]  # procent relativ (1990 = 100%)

plt.figure(figsize=(8,5))
plt.plot(years, emissions, marker='o')

plt.title("Traiectoria reducerii emisiilor în UE")
plt.xlabel("An")
plt.ylabel("Nivel emisii (% față de 1990)")

plt.grid(True)
plt.show()
```

Graficul ilustrează traiectoria planificată de reducere a emisiilor de gaze cu efect de seră la nivelul Uniunii Europene. Se observă un obiectiv ambițios de reducere semnificativă până în anul 2030, urmat de atingerea neutralității climatice până în 2050. Această evoluție reflectă angajamentul UE de a deveni lider global în combaterea schimbărilor climatice.

Germania este una dintre țările europene care au un rol central în promovarea și finanțarea inițiativelor climatice. Un exemplu relevant este International Climate Initiative (IKI), inițiativă susținută de guvernul german, care finanțează proiecte privind clima, biodiversitatea și tranziția verde. În paralel, profilul de țară publicat de European Environment Agency arată că Germania înregistrează progrese importante în anumite domenii, dar și decalaje în altele, ceea ce o face un exemplu de lider climatic „complex”: avansat instituțional, dar încă în proces de transformare structurală.

### Suedia — lider legislativ în politici climatice

Suedia este una dintre primele țări care au implementat politici climatice bazate pe instrumente economice și legislative coerente.

Un element central îl reprezintă introducerea taxei pe carbon în 1991, una dintre primele de acest tip la nivel global. În prezent, Suedia aplică una dintre cele mai ridicate taxe pe carbon, contribuind semnificativ la reducerea emisiilor de CO₂.

În 2017, Suedia a adoptat un cadru legislativ climatic care stabilește obiective pe termen lung, inclusiv atingerea neutralității climatice până în 2045. Acest cadru este susținut de instituții independente care monitorizează implementarea politicilor.

Eficiența acestor măsuri este reflectată în reducerea emisiilor per capita și a intensității carbonice a economiei, ambele situându-se sub media Uniunii Europene. De asemenea, sistemele de taxare a carbonului și EU ETS acoperă majoritatea emisiilor provenite din combustibili fosili.

Aceste elemente poziționează Suedia ca un exemplu de bună practică în domeniul politicilor climatice la nivel european.

### Danemarca — lider în guvernanță climatică

Danemarca se remarcă printr-un model avansat de guvernanță climatică, bazat pe legislație strictă și mecanisme instituționale solide.

Legea climatică adoptată în 2020 stabilește obiective obligatorii de reducere a emisiilor până în 2030 și introduce un mecanism de revizuire periodică a politicilor climatice, evaluat de un consiliu independent de experți.

Un element inovator îl reprezintă introducerea unei taxe pe emisiile de CO₂ în sectorul agricol, considerată o premieră la nivel global. Aceasta extinde politicile climatice către un sector dificil de reglementat.

Reforma fiscală verde din 2022 a accelerat tranziția industrială, prin stimularea investițiilor în tehnologii curate și reducerea emisiilor.

La nivel internațional, Danemarca participă la proiecte de captare și stocare a carbonului (CCS), în colaborare cu alte state europene, contribuind la dezvoltarea soluțiilor tehnologice pentru reducerea emisiilor.

Prin aceste măsuri, Danemarca este considerată un model de implementare eficientă a politicilor climatice în Uniunea Europeană.

```python
import pandas as pd
import plotly.express as px
import plotly.io as pio
pio.renderers.default = "vscode"
# citire date EPI din CSV local
df_epi = pd.read_csv(r"D:\Downloads\ICI proiect\ICI_Proiect\data\raw\epi_2024.csv")

fig = px.choropleth(
    df_epi,
    locations="Country",
    locationmode="country names",
    color="EPI",
    hover_name="Country",
    color_continuous_scale=[
        [0.0, "#efe6a0"],
        [0.2, "#c8de9b"],
        [0.4, "#8fd0b2"],
        [0.6, "#4cb5c4"],
        [0.8, "#2f7fbe"],
        [1.0, "#102f84"]
    ],
    title="Environmental Performance Index (EPI) "
)

fig.update_layout(
    width=1000,
    height=650,
    margin=dict(l=10, r=10, t=60, b=10),
    coloraxis_colorbar=dict(title="EPI Score")
)
fig.write_html("../outputs/figures/epi_world_map.html")
fig.show()
```
